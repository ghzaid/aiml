```python
# Road network represented using an adjacency list
graph = {
    'A': ['B', 'D'],
    'B': ['A', 'C'],
    'C': ['B', 'E'],
    'D': ['A'],
    'E': ['C'],
    'F': ['G'],
    'G': ['F']
}


# DFS to check whether a route exists
def route_exists(graph, start, destination, visited=None):

    if visited is None:
        visited = set()

    if start == destination:
        return True

    visited.add(start)

    for neighbour in graph[start]:
        if neighbour not in visited:
            if route_exists(graph, neighbour, destination, visited):
                return True

    return False


# DFS to find one connected component
def find_component(graph, start, visited):
    component = []

    visited.add(start)
    component.append(start)

    for neighbour in graph[start]:
        if neighbour not in visited:
            component.extend(
                find_component(graph, neighbour, visited)
            )

    return component


# Find all connected components
def connected_components(graph):
    visited = set()
    components = []

    for city in graph:
        if city not in visited:
            component = find_component(graph, city, visited)
            components.append(component)

    return components


# Test route checking
start_city = 'A'
destination_city = 'E'

if route_exists(graph, start_city, destination_city):
    print(f"Route exists from {start_city} to {destination_city}")
else:
    print(f"No route exists from {start_city} to {destination_city}")


# Find connected components
components = connected_components(graph)

print("\nConnected Components:")

for i, component in enumerate(components, 1):
    print(f"Component {i}: {component}")
```
