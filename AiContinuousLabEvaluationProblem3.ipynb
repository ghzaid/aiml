```python
from collections import deque


def water_jug_bfs():
    capacity_a = 4
    capacity_b = 3
    target = 2

    start = (0, 0)

    queue = deque([start])
    visited = {start}

    parent = {start: None}
    action = {start: "Start"}

    while queue:
        a, b = queue.popleft()

        # Check whether target is reached
        if a == target or b == target:
            path = []

            current = (a, b)

            while current is not None:
                path.append((current, action[current]))
                current = parent[current]

            path.reverse()
            return path

        # Generate possible states

        states = [
            ((capacity_a, b), "Fill Jug A"),
            ((a, capacity_b), "Fill Jug B"),
            ((0, b), "Empty Jug A"),
            ((a, 0), "Empty Jug B")
        ]

        # Pour A -> B
        amount = min(a, capacity_b - b)
        states.append(
            ((a - amount, b + amount), "Pour Jug A into Jug B")
        )

        # Pour B -> A
        amount = min(b, capacity_a - a)
        states.append(
            ((a + amount, b - amount), "Pour Jug B into Jug A")
        )

        # Add unvisited states
        for new_state, operation in states:
            if new_state not in visited:
                visited.add(new_state)
                parent[new_state] = (a, b)
                action[new_state] = operation
                queue.append(new_state)

    return None


# Solve the problem
solution = water_jug_bfs()

print("Minimum sequence of steps:\n")

for i, (state, operation) in enumerate(solution):
    print(f"{i}. {operation:25} State = {state}")
```
